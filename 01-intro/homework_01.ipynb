#### Sum of weights

```python
# 1–3. Select Asian cars, two columns, and the first 7 rows
asian_cars = df[df['origin'] == 'Asia']
selected_cars = asian_cars[['vehicle_weight', 'model_year']].head(7)

# 4. Convert the DataFrame into a NumPy array
X = selected_cars.to_numpy()

# 5. Multiply the transpose of X by X
XTX = X.T @ X

# 6. Invert XTX
XTX_inverse = np.linalg.inv(XTX)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Calculate w
w = XTX_inverse @ X.T @ y

# 9. Sum the elements of w
result = w.sum()

print("X:")
print(X)

print("\nw:")
print(w)

print("\nSum:", result)
```
